# Metódusok

### Eljárás, paraméter nélkül

```python
def koszones():
     print("Üdvözöllek")
```

```python
koszones()
```

### Eljárás, paraméterrel

```python
def udvozles(nev):
     print(f"Üdvözöllek: {nev}")
```

```python
udvozles("Sanyi")
```

### Függvény, paraméter nélkül

```python
def func_koszonoes():
     return "Üdvözöllek"
```

```python
print(func_koszonoes())
```

### Függvény, paraméterrel

```python
def func_udvozles(nev):
     return f"Üdvözöllek: {nev}"
```

```python
print(func_udvozles("Sanyi"))
```
